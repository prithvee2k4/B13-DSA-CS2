```python
import pandas as pd

literacy_date_path = "/content/drive/MyDrive/DSAB13/CaseStudy/CS1/literacy_rate.xml"

literacy_date_df = pd.read_xml(literacy_date_path)

print("Literacy Rate Dataset:")
display(literacy_date_df)
print("\n")

print("First 5 rows:")
display(literacy_date_df.head())
print("\n")

print("\nColumn Names:")
print(literacy_date_df.columns)
print("\n")

print("\nData Types:")
print(literacy_date_df.dtypes)
print("\n")

#PART B- TASK 5: To explore datasets.

print("No : of rows:")
print(literacy_date_df.shape[0]) #No : of rows
print("\n")

print("No : of columns:")
print(literacy_date_df.shape[1]) #No : of columns
print("\n")

print("Column Names:")
print(literacy_date_df.columns) #Column Names
print("\n")

print("Data Types:")
print(literacy_date_df.dtypes) #Data Types
print("\n")

print("Missing Values:")
print(literacy_date_df.isnull().sum()) #Missing Values
print("\n")

print("Duplicate records")
print(literacy_date_df.duplicated().sum()) #Duplicate records
print("\n")
```

```python
#TASK 6
# To fetch and check column names

print("Population columns:")
print(population_df.columns)

print("\nGDP columns:")
print(gdp_df.columns)

print("\nInternet Users columns:")
print(internet_users_df.columns)

print("\nLiteracy Rate columns:")
print(literacy_date_df.columns)
```

```python
# Rename only 2 columns.
gdp_df.rename(columns={'Country': 'country'}, inplace=True)

literacy_date_df.rename(columns={'name': 'country'}, inplace=True)

# To verify
print(population_df.columns)
print(gdp_df.columns)
print(internet_users_df.columns)
print(literacy_date_df.columns)
```

```python
# TASK 7
# Check datatypes
print("Population:")
print(population_df.dtypes)

print("\nGDP:")
print(gdp_df.dtypes)

print("\nInternet Users:")
print(internet_users_df.dtypes)

print("\nLiteracy Rate:")
print(literacy_date_df.dtypes)


print("\n")

#To change to appropriate datatypes

population_df['population'] = pd.to_numeric(population_df['population'])
gdp_df['GDP'] = pd.to_numeric(gdp_df['GDP'])
internet_users_df['internet_users'] = pd.to_numeric(internet_users_df['internet_users'])
literacy_date_df['literacy_rate'] = pd.to_numeric(literacy_date_df['literacy_rate'])
```

```python
# TASK 8
# to merge all 4 datasets into 1 df. named = 'merged_df'

merged_df = pd.merge(population_df, gdp_df, on='country')
merged_df = pd.merge(merged_df, internet_users_df, on='country')
merged_df = pd.merge(merged_df, literacy_date_df, on='country')

# To display
print(merged_df)
```

```python
# TASK 9
# 1. Internet Penetration (%)
# The formula is:
# Internet Penetration = (Internet Users / Population) × 100
merged_df['Internet Penetration (%)'] = (
    merged_df['internet_users'] / merged_df['population'] * 100
).round(2)

# 2. GDP Per Capital
# The formula is:
# GDP Per Capita = GDP / Population

merged_df['GDP Per Capital'] = (
    merged_df['GDP'] * 1000000 / merged_df['population']
).round(2)

# To display the merged_df

display(merged_df)
```

```python
#PART D
#TASK 10

import sqlite3

# Create SQLite database
conn = sqlite3.connect("country_statistics.db")

# Store DataFrame in SQLite table
merged_df.to_sql("country_statistics", conn, if_exists="replace", index=False)

print("Data successfully stored in country_statistics table.")

# Checking whether the data was stored in table country_statistics

result = pd.read_sql("SELECT * FROM country_statistics", conn)

display(result)
```

```python
# TASK 11
# To Retrieve the first five records from the SQLite table.

query = "SELECT * FROM country_statistics LIMIT 5"

result = pd.read_sql(query, conn)

display(result)
```

```python
# PART E

# 1. Top 5 countries by GDP
print("1. Top 5 Countries by GDP")
display(merged_df.nlargest(5, 'GDP')[['country', 'GDP']])
print("\n")

# 2. Top 5 countries by population
print("2. Top 5 Countries by Population")
display(merged_df.nlargest(5, 'population')[['country', 'population']])
print("\n")

# 3. Average literacy rate
print("3. Average Literacy Rate:")
print(round(merged_df['literacy_rate'].mean(), 2))
print("\n")

# 4. Literacy rate greater than 90%
print("\n4. Countries with Literacy Rate > 90%")
display(merged_df[merged_df['literacy_rate'] > 90][['country', 'literacy_rate']])
print("\n")

# 5. Internet penetration greater than 70%
print("\n5. Countries with Internet Penetration > 70%")
display(merged_df[merged_df['Internet Penetration (%)'] > 70][['country', 'Internet Penetration (%)']])
print("\n")

# 6. Top 5 countries by GDP per capita
print("\n6. Top 5 Countries by GDP Per Capital")
display(merged_df.nlargest(5, 'GDP Per Capital')[['country', 'GDP Per Capita']])
print("\n")

# 7. Total population
print("\n7. Total Population:")
print(merged_df['population'].sum())
print("\n")

# 8. Average internet penetration
print("\n8. Average Internet Penetration:")
print(round(merged_df['Internet Penetration (%)'].mean(), 2))
print("\n")

# 9. Highest number of internet users
print("\n9. Country with Highest Internet Users:")
display(merged_df.loc[merged_df['internet_users'].idxmax(), ['country', 'internet_users']])
print("\n")

# 10. Correlation
print("\n10. Correlation between Literacy Rate and Internet Penetration:")
print(round(merged_df['literacy_rate'].corr(merged_df['Internet Penetration (%)']), 2))
print("\n")

```

```python
# Q11 - Display the Top 5 countries by GDP
q11 = """
SELECT country, GDP
FROM countries
ORDER BY GDP DESC
LIMIT 5;
"""

print("Q11 - Top 5 Countries by GDP")
display(pd.read_sql_query(q11, conn))
print("\n")


# Q12 - Find the average literacy rate
q12 = """
SELECT AVG(literacy_rate) AS Average_Literacy_Rate
FROM countries;
"""

print("Q12 - Average Literacy Rate")
display(pd.read_sql_query(q12, conn))
print("\n")


# Q13 - Display all countries having literacy rate above 90%
q13 = """
SELECT country, literacy_rate
FROM countries
WHERE literacy_rate > 90;
"""

print("Q13 - Countries with Literacy Rate Above 90%")
display(pd.read_sql_query(q13, conn))
print("\n")


# Q14 - Find the Top 5 countries with the highest internet penetration
q14 = """
SELECT country, "Internet Penetration (%)"
FROM countries
ORDER BY "Internet Penetration (%)" DESC
LIMIT 5;
"""

print("Q14 - Top 5 Countries by Internet Penetration")
display(pd.read_sql_query(q14, conn))
print("\n")


# Q15 - Calculate the total population
q15 = """
SELECT SUM(population) AS Total_Population
FROM countries;
"""

print("Q15 - Total Population")
display(pd.read_sql_query(q15, conn))
print("\n")


# Q16 - Find the country with the highest GDP per capita
q16 = """
SELECT country, "GDP Per Capita"
FROM countries
ORDER BY "GDP Per Capita" DESC
LIMIT 1;
"""

print("Q16 - Country with Highest GDP Per Capita")
display(pd.read_sql_query(q16, conn))
print("\n")


# Q17 - Export the final merged dataset as a CSV file
final_df = pd.read_sql_query("SELECT * FROM countries", conn)

final_df.to_csv("final_merged_dataset.csv", index=False)

print("Q17 - Final merged dataset exported successfully")
print("\n")


# Q18 - Sort the dataset in descending order of GDP Per Capita
# and display the Top 10 countries
q18 = """
SELECT country, "GDP Per Capita"
FROM countries
ORDER BY "GDP Per Capita" DESC
LIMIT 10;
"""

print("Q18 - Top 10 Countries by GDP Per Capita")
display(pd.read_sql_query(q18, conn))
print("\n")
```
